```python
# Recursion Task 1
def sum_nested_list(nested_list):
  total = 0
  for element in nested_list:
    if isinstance(element, list):
      total += sum_nested_list(element)
    else:
      total += element
  return total

nested_list = ([1, [2, [3, 4], 5], 6, [7, 8]])
result = sum_nested_list(nested_list)
print("Sum of Nested List:", result)
```

```python
def generate_permutations(s):

  if len(s) == 1:
    return [s]

  permutations = []
  seen = set()

  for i in range(len(s)):
    char = s[i]

    if char in seen:
      continue
    seen.add(char)

    remaining = s[:i] + s[i+1:]


    for perm in generate_permutations(remaining):
      permutations.append(char + perm)

  return permutations


print("Permutations of 'abc':")
print(generate_permutations("abc"))

print("\nPermutations of 'aab':")
print(generate_permutations("aab"))





```

```python
def min_coins(coins, amount):
    """
    Returns the minimum number of coins needed to make the given amount.
    Uses Dynamic Programming (Tabulation).
    """
    # Initialize DP array: dp[i] = minimum coins to make amount i
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0  # Base case

    for coin in coins:
        for i in range(coin, amount + 1):
            dp[i] = min(dp[i], dp[i - coin] + 1)

    return dp[amount] if dp[amount] != float('inf') else -1
print(min_coins([1, 2, 5], 11))
```

```python
def longest_common_subsequence(s1, s2):
    """
    Returns the length of the Longest Common Subsequence between s1 and s2.
    Uses DP with a 2D table.
    """
    n, m = len(s1), len(s2)

    # Create DP table of size (n+1) x (m+1)
    dp = [[0] * (m + 1) for _ in range(n + 1)]

    # Fill table
    for i in range(1, n + 1):
        for j in range(1, m + 1):
            if s1[i - 1] == s2[j - 1]:  # Characters match
                dp[i][j] = dp[i - 1][j - 1] + 1
            else:  # No match — take max from top or left
                dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])

    return dp[n][m]

print(longest_common_subsequence("abcde", "ace"))
```

```python
def knapsack(weights, values, capacity):
    """
    Returns the maximum value achievable with the given weight capacity.
    Uses DP for 0/1 Knapsack problem.
    """
    n = len(weights)

    # DP table: (n+1) x (capacity+1)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        for w in range(1, capacity + 1):
            if weights[i - 1] <= w:
                # Choose max: take the item OR leave it
                dp[i][w] = max(values[i - 1] + dp[i - 1][w - weights[i - 1]],
                               dp[i - 1][w])
            else:
                dp[i][w] = dp[i - 1][w]

    return dp[n][capacity]

weights = [1, 3, 4, 5]
values = [1, 4, 5, 7]
capacity = 7

print(knapsack(weights, values, capacity))
```
